```python
# ===============================
# AI Web Scraping Agent (Fixed)
# Target: books.toscrape.com
# Extract: Books under £20
# Output: CSV + JSON
# ===============================

import requests
from bs4 import BeautifulSoup
import pandas as pd
import time

# Base URL
BASE_URL = "http://books.toscrape.com/catalogue/page-{}.html"

books_data = []

# --------------------------------
# Function to scrape one page
# --------------------------------
def scrape_page(page_number):

    url = BASE_URL.format(page_number)

    try:
        response = requests.get(url, timeout=10)

        if response.status_code != 200:
            return False   # Stop pagination

        soup = BeautifulSoup(response.text, "html.parser")

        books = soup.find_all("article", class_="product_pod")

        if not books:
            return False

        for book in books:
            try:
                # Title
                title = book.h3.a["title"]

                # Price cleaning
                price_text = book.find("p", class_="price_color").text
                price = float(price_text.replace("£", "").replace("Â", ""))

                # Availability
                availability = book.find(
                    "p", class_="instock availability"
                ).get_text(strip=True)

                # Filter books under £20
                if price < 20:
                    books_data.append({
                        "Title": title,
                        "Price (£)": price,
                        "Availability": availability
                    })

            except Exception as e:
                print("Tag missing:", e)

        return True

    except requests.exceptions.RequestException as e:
        print("Request error:", e)
        return False


# --------------------------------
# Pagination Loop
# --------------------------------
page = 1

while True:
    print(f"Scraping page {page}...")

    if not scrape_page(page):
        break

    page += 1
    time.sleep(1)  # Ethical delay


# --------------------------------
# Save Results
# --------------------------------
df = pd.DataFrame(books_data)

df.to_csv("books_under_20.csv", index=False)
df.to_json("books_under_20.json", orient="records", indent=4)

print("✅ Scraping Complete")
print("Total books:", len(df))

df.head()
```
